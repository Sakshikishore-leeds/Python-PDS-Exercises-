```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df_titanic = pd.read_csv("/content/train (4).csv")
```

```python
df_missing_age = df_titanic['Age'].isnull().sum()
df_missing_cabin = df_titanic['Cabin'].isnull().sum()
df_missing_embarked = df_titanic['Embarked'].isnull().sum()
print(df_missing_age)
print(df_missing_cabin)
print(df_missing_embarked)
""" There are totally 177 values missing in the age column and 687 values missing in the cabin column and 2 missing values in the embarked column"""
```

```python
df_clean = df_titanic.dropna(subset=['Embarked'])
```

```python
plt.figure()
df_clean['Embarked'].value_counts().plot.pie(autopct='%1.1f%%')
plt.title('Percentage of Passengers by Embarktion Port')
plt.ylabel('')
plt.show()

"""  S - 72.4%  q  = 8.7% and C - 18.9%"""
```

```python
plt.figure()
sns.countplot(x='Pclass',data=df_clean)
plt.title('Passengers Class Distribution in Titanic')
plt.xlabel('Number of Passengers')
plt.ylabel('Class')
plt.show()
"""   3rd class has the most passengers"
```

```python
for pclass in [1,2,3]:
  plt.figure()
  data = df_titanic[(df_titanic['Sex'] == 'female') & (df_titanic['Pclass'] == pclass)]
  sns.countplot(x='Survived',data= data)
  plt.title(f'Female Passengers Survival by Class{pclass}')
  plt.xlabel('Survival')
  plt.ylabel('Count')
  plt.show()
```

```python
plt.figure(figsize(12,5))
plt.subplot(1,2,1)
sns.boxplot(x='Survived',y='Age',data=df_titanic['Sex']=='female')
```
